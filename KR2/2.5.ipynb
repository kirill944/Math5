## Решение

Комплексная постоянная:
$$
B=u+iv
$$
Тогда сопряженное:
$$
\bar{B}=u−iv
$$
$$
\bar{B}z + B\bar{z} = 2ux + 2vy
$$

Подставляем:
$$
A(x^2 + y^2) + 2ux + 2vy + C = 0
$$

Делим на A и выделяем полные квадраты:
$$
x^2 + y^2 + 2ux/A + 2uy/A + C/A = 0
$$
$$
x^2 + 2ux/A + u^2/A^2 + y^2 + 2vy/A + v^2/A^2 + C/A = u^2/A^2 + v^2/A^2
$$
$$
(x + u/A)^2 + (y + v/A)^2 = (u^2 + v^2 − AC) / A^2
$$
## Ответ

Центр:
$$
(-u/A; -v/A) = -B/A
$$
Радиус:
$$
R = \sqrt{(u^2 + v^2 − AC) / A^2} = \sqrt{|B|² − AC)} / A
$$
Условие AC < |B|² гарантирует R² > 0, т.е. радиус действителен.

```python
import sympy as sp
# 1. Определяем символьные переменные
# A и C - действительные числа
A, C = sp.symbols('A C', real=True, positive=True)
# u и v - действительная и мнимая части комплексного числа B
u, v = sp.symbols('u v', real=True)
# x и y - действительная и мнимая части z
x, y = sp.symbols('x y', real=True)

# 2. Вводим выражения для z, B и сопряженных
z = x + sp.I * y
B = u + sp.I * v
B_conj = u - sp.I * v
z_conj = x - sp.I * y

# 3. Записываем исходное выражение левой части уравнения
expr = A * sp.Abs(z)**2 + B_conj * z + B * z_conj + C

# 4. Раскрываем модуль и комплексные числа
# sp.expand помогает раскрыть скобки
expr_expanded = sp.expand(expr)

# 5. Приводим подобные и собираем по x и y, чтобы выделить полный квадрат
# Упрощаем: I^2 = -1
expr_simplified = sp.simplify(expr_expanded)

# Разделим выражение на A, чтобы привести к каноническому виду окружности
expr_divided = expr_simplified / A

# 6. Выделяем полные квадраты вручную через символьные преобразования
# Нам нужно получить вид: (x - x0)^2 + (y - y0)^2 = R^2
# Для этого найдем x0 и y0 из линейных членов (2*u*x/A и -2*v*y/A)

# Собираем коэффициенты при x и y
coeff_x = sp.expand(expr_divided).coeff(x)
coeff_y = sp.expand(expr_divided).coeff(y)

# Центр окружности (x0, y0) находится как -coeff / 2
x0 = -coeff_x / 2
y0 = -coeff_y / 2

# 7. Вычисляем правую часть уравнения (радиус в квадрате)
# Подставляем найденные x0 и y0 в уравнение и переносим константу
# Уравнение: (x - x0)^2 + (y - y0)^2 = R^2
# Найдем R^2, вычтя квадраты смещений и приравняв к константе
R_squared = sp.simplify(x0**2 + y0**2 - (expr_divided.subs({x:0, y:0})))
R = sp.sqrt(R_squared)

# Выводим результаты
print(f"Уравнение окружности: (x - ({x0}))^2 + (y - ({y0}))^2 = {R_squared}")
print(f"\nЦентр окружности: z0 = {sp.simplify(x0)} + i * ({sp.simplify(y0)})")
print(f"Радиус окружности: R = {sp.simplify(R)}")

# Проверка условия AC < |B|^2 для положительности радиуса
# |B|^2 = u^2 + v^2
print(f"\nПроверка условия AC < |B|^2 (подкоренное выражение > 0):")
print(f"Подкоренное выражение для R: (u^2 + v^2 - A*C) / A^2")
print("Так как по условию AC < |B|^2, то u^2 + v^2 - A*C > 0, следовательно R > 0.")
```
