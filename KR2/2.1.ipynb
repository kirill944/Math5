## Основные определения

### Комплексное число

**Комплексное число** записывается в алгебраической форме

$$
z=x+iy,
$$

где:

- $x=\operatorname{Re}(z)$ — действительная часть;
- $y=\operatorname{Im}(z)$ — мнимая часть;
- $i$ — мнимая единица, для которой

$$
i^2=-1.
$$

---

### Комплексная плоскость

Комплексному числу

$$
z=x+iy
$$

соответствует точка $(x,y)$ на плоскости:

- горизонтальная ось — $\operatorname{Re}(z)$;
- вертикальная ось — $\operatorname{Im}(z)$.

---

### Модуль комплексного числа

Для

$$
z=x+iy
$$

модуль определяется формулой

$$
|z|=\sqrt{x^2+y^2}.
$$

Модуль — это расстояние от начала координат до точки $z$ на комплексной плоскости.

---

### Тригонометрическая форма комплексного числа

Любое ненулевое комплексное число можно записать в виде

$$
z=r(\cos\varphi+i\sin\varphi),
$$

где:

- $r=|z|$ — модуль;
- $\varphi=\arg z$ — аргумент.

---

### Показательная форма

По формуле Эйлера

$$
e^{i\varphi}=\cos\varphi+i\sin\varphi.
$$

Поэтому тригонометрическую форму можно записать короче:

$$
z=re^{i\varphi}.
$$

---

### Формула Муавра

Для комплексного числа

$$
z=r(\cos\varphi+i\sin\varphi)
$$

выполняется

$$
z^n=
r^n
\left(
\cos(n\varphi)+i\sin(n\varphi)
\right).
$$

Эта формула позволяет возводить комплексные числа в целую степень.

Для извлечения корней используется обратная идея.

---

### Формула корней $n$-й степени

Пусть

$$
w=R(\cos\theta+i\sin\theta).
$$

Решения уравнения

$$
z^n=w
$$

имеют вид

$$
z_k=
\sqrt[n]{R}
\left(
\cos\frac{\theta+2\pi k}{n}
+
i\sin\frac{\theta+2\pi k}{n}
\right),
$$

где

$$
k=0,1,\ldots,n-1.
$$

Получается ровно $n$ различных корней.

Все они:

- имеют одинаковый модуль;
- лежат на одной окружности с центром в начале координат;
- расположены на равных угловых расстояниях друг от друга.

Расстояние по углу между соседними корнями равно

$$
\frac{2\pi}{n}.
$$

```python
import sympy as sp
import math


def forms(z_exact):
    z = sp.nsimplify(sp.simplify(sp.expand_complex(z_exact)))
    re, im = sp.re(z), sp.im(z)
    r = sp.simplify(sp.sqrt(re ** 2 + im ** 2))
    phi = sp.simplify(sp.arg(z))
    print('Алгебраическая:     ', sp.simplify(z))
    print('Модуль r =', r, '≈', float(r))
    print('Аргумент φ =', phi, '≈', float(phi), f'({math.degrees(float(phi)):.4f}°)')
    print('Тригонометрическая: ', f'{r}*(cos({phi}) + i*sin({phi}))')
    print('Показательная:      ', f'{r}*exp(i*({phi}))')


I = sp.I
```

```python
z_a = (5 + I) * (7 - 6 * I) / (3 + I)
forms(z_a)
```

```python
z_b = (1 + I) ** 5 / (1 - I) ** 3
forms(z_b)
```

```python
z_c = (-1 + I * sp.sqrt(3)) ** 15 / (1 - I) ** 20 + (-1 - I * sp.sqrt(3)) ** 15 / (1 + I) ** 20
forms(z_c)
```
