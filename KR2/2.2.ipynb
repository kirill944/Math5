# Решение

## Шаг 1. Представляем правую часть в тригонометрической форме

Имеем

$$
1+i.
$$

Находим модуль:

$$
R=|1+i|=\sqrt2.
$$

Аргумент:

$$
\varphi=\frac{\pi}{4}.
$$

Поэтому

$$
1+i=
\sqrt2
\left(
\cos\frac{\pi}{4}
+
i\sin\frac{\pi}{4}
\right).
$$

---

## Шаг 2. Представляем неизвестное число в тригонометрической форме

Пусть

$$
z=r(\cos\varphi+i\sin\varphi).
$$

Тогда по формуле Муавра

$$
z^8=
r^8
\left(
\cos 8\varphi+i\sin 8\varphi
\right).
$$

По условию

$$
z^8=1+i.
$$

Следовательно,

$$
r^8
\left(
\cos 8\varphi+i\sin 8\varphi
\right)
=
\sqrt2
\left(
\cos\frac{\pi}{4}
+
i\sin\frac{\pi}{4}
\right).
$$

---

## Шаг 3. Находим модуль корней

Сравниваем модули:

$$
r^8=\sqrt2.
$$

Следовательно, все восемь корней имеют одинаковый модуль

$$
\boxed{r=2^{1/16}}.
$$

Численно:

$$
2^{1/16}\approx1.04427.
$$

---

## Шаг 4. Находим аргументы корней

Аргументы комплексных чисел могут отличаться на целое число полных оборотов, поэтому

$$
8\varphi=
\frac{\pi}{4}+2\pi k,
$$

где $k\in\mathbb Z$.

Делим на $8$:

$$
\varphi=
\frac{\pi}{32}+\frac{\pi k}{4}.
$$

Для получения всех различных корней достаточно взять

$$
k=0,1,2,\ldots,7.
$$

Значит углы корней:

$$
\varphi_k=
\frac{\pi}{32}+\frac{\pi k}{4}.
$$

Соседние корни отличаются по аргументу на

$$
\frac{\pi}{4}=45^\circ.
$$

---

## Шаг 5. Записываем все решения

Получаем формулу

$$
\boxed{
z_k=
2^{1/16}
\left(
\cos\left(\frac{\pi}{32}+\frac{\pi k}{4}\right)
+
i\sin\left(\frac{\pi}{32}+\frac{\pi k}{4}\right)
\right)
}
$$

для

$$
\boxed{k=0,1,\ldots,7}.
$$

В показательной форме:

$$
\boxed{
z_k=
2^{1/16}
e^{\,i\left(\frac{\pi}{32}+\frac{\pi k}{4}\right)},
\qquad
k=0,\ldots,7.
}
$$

---

## Шаг 6. Геометрический смысл решения

Все восемь корней имеют одинаковый модуль

$$
|z_k|=2^{1/16}.
$$

Поэтому они лежат на окружности

$$
|z|=2^{1/16}
$$

с центром в начале координат.

Корни расположены равномерно по окружности, поскольку разность аргументов соседних корней равна

$$
\frac{\pi}{4}.
$$

Первый корень имеет аргумент

$$
\frac{\pi}{32},
$$

то есть

$$
\frac{180^\circ}{32}=5.625^\circ.
$$

Остальные получаются последовательным поворотом на $45^\circ$.

---

## Шаг 7. Проверка

Для каждого найденного корня должно выполняться

$$
z_k^8=1+i.
$$

В программе мы проверим это численно.

Из-за вычислений с числами с плавающей точкой может возникать очень маленькая погрешность, поэтому вместо точного сравнения `==` используем `np.allclose`.

---

## Ответ

Уравнение

$$
z^8=1+i
$$

имеет восемь решений:

$$
\boxed{
z_k=
2^{1/16}
\left(
\cos\left(\frac{\pi}{32}+\frac{\pi k}{4}\right)
+
i\sin\left(\frac{\pi}{32}+\frac{\pi k}{4}\right)
\right),
\quad
k=0,\ldots,7.
}
$$

Все решения лежат на окружности

$$
\boxed{|z|=2^{1/16}}
$$

и расположены на ней равномерно с угловым шагом

$$
\boxed{\frac{\pi}{4}=45^\circ}.
$$


```python
import numpy as np
import matplotlib.pyplot as plt

print("Решаем уравнение z^8 = 1 + i\n")

# ШАГ 1. Задаём правую часть уравнения

w = 1 + 1j
print("Правая часть w =", w)


# ШАГ 2. Находим модуль и аргумент числа 1 + i

R = abs(w)
theta = np.angle(w)

print("\nМодуль |1 + i| =", R)
print("Аргумент arg(1 + i) =", theta, "радиан")
print("Аргумент в долях pi =", theta / np.pi, "* pi")


# ШАГ 3. Находим модуль каждого корня

root_radius = R ** (1 / 8)

print("\nМодуль каждого корня:")
print("r = |w|^(1/8) =", root_radius)
print("2^(1/16) =", 2 ** (1 / 16))

# ШАГ 4. Находим аргументы всех восьми корней

k = np.arange(8)

angles = (theta + 2 * np.pi * k) / 8

print("\nАргументы корней:")
for index, angle in enumerate(angles):
    print(
        f"k = {index}: "
        f"phi_{index} = {angle:.9f} рад = "
        f"{np.degrees(angle):.3f}°"
    )

# ШАГ 5. Строим сами комплексные корни

roots = root_radius * np.exp(1j * angles)

print("\nОТВЕТ: восемь корней в алгебраической форме\n")

for index, z in enumerate(roots):
    print(
        f"z_{index} = "
        f"{z.real:.9f} "
        f"{z.imag:+.9f}i"
    )

# ШАГ 6. Проверяем найденные решения

check_values = roots ** 8
check = np.allclose(check_values, w)

print("\nПроверка всех корней:")
print("Для каждого z_k выполняется z_k^8 = 1 + i:", check)

# ШАГ 7. Изображаем решения на комплексной плоскости

x = roots.real
y = roots.imag

# Создаём рисунок.
plt.figure(figsize=(8, 8))
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
t = np.linspace(0, 2 * np.pi, 500)

circle_x = root_radius * np.cos(t)
circle_y = root_radius * np.sin(t)

plt.plot(
    circle_x,
    circle_y,
    linestyle="--",
    label=r"$|z|=2^{1/16}$"
)

# Наносим на комплексную плоскость восемь найденных корней.
plt.scatter(
    x,
    y,
    s=70,
    zorder=3
)
for index, z in enumerate(roots):
    plt.annotate(
        f"$z_{index}$",
        (z.real, z.imag),
        xytext=(8, 8),
        textcoords="offset points"
    )

# Делаем одинаковый масштаб по обеим осям,
# чтобы окружность действительно выглядела окружностью.
plt.gca().set_aspect("equal", adjustable="box")

# Добавляем подписи.
plt.title(r"Решения уравнения $z^8=1+i$")
plt.xlabel(r"$\operatorname{Re}(z)$")
plt.ylabel(r"$\operatorname{Im}(z)$")
plt.grid(True, alpha=0.3)
plt.legend()
limit = root_radius * 1.25
plt.xlim(-limit, limit)
plt.ylim(-limit, limit)

plt.show()
```
