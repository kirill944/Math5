
### Важные для нас свойства:

Свойство разности
$$arg(\frac{z - z_1}{z - z_2}) = arg(z - z_1) - arg(z - z_2)$$
Геометрически это угол между векторами, идущими из точки $z$ к точкам $z_1$ и $z_2$.

Следующее уравнение
$$arg(\frac{z - z_1}{z - z_2}) = \theta$$
задаёт дугу окружности, проходящую через $z_1$ и $z_2$. Из любой точки этой дуги отрезок $z_1z_2$ виден под одним и тем же углом $\theta$.


Двойное неравенство $A < arg(\frac{z - z_1}{z - z_2}) < B$ задаёт область, ограниченную двумя дугами окружностей (своеобразную "линзу")

Для решения задач такого типа нужно перевести алгебраические условия на язык геометрии. Основной метод: обозначить $z = x + iy$, где $x = \text{Re}(z)$, $y = \text{Im}(z)$, и посмотреть, какое множество точек $(x, y)$ на плоскости описывает неравенство или уравнение.

Вот подробный разбор каждого пункта:

### Общий алгоритм:
1.  Подставить $z = x + iy$.
2.  Упростить выражение (раскрыть скобки, выделить действительную и мнимую части, найти модули).
3.  Вспомнить геометрический смысл полученного уравнения/неравенства.
4.  Нарисовать эскиз.

### а) $|z - i| + |z + i| < 4$
*   **Геометрический смысл:** Сумма расстояний от точки $z$ до точек $i$ (координата $(0, 1)$) и $-i$ (координата $(0, -1)$) меньше 4.
*   **Что это:** Это эллипс с фокусами в точках $(0, 1)$ и $(0, -1)$. Большая полуось $a = 2$ (так как сумма расстояний $2a = 4$). Фокальное расстояние $c = 1$. Малая полуось $b = \sqrt{a^2 - c^2} = \sqrt{4 - 1} = \sqrt{3}$.

### b) $\text{Re}(z(1 - i)) < \sqrt{2}$
*   **Алгебра:** Пусть $z = x + iy$. Тогда $z(1 - i) = (x + iy)(1 - i) = x - ix + iy - i^2y = (x + y) + i(y - x)$.
*   Действительная часть: $\text{Re} = x + y$.
*   **Неравенство:** $x + y < \sqrt{2} \Rightarrow y < -x + \sqrt{2}$.
*   **Что это:** Полуплоскость ниже прямой $y = -x + \sqrt{2}$.
*   **Как рисовать:** Прямая проходит через точки $(\sqrt{2}, 0)$ и $(0, \sqrt{2})$.

### c) $|z + 1| = |z + i| = |z - i|$
*   **Геометрический смысл:** Точка $z$ равноудалена от трех точек: $-1$, $-i$ и $i$.
*   **Алгебра:** Это центр описанной окружности треугольника с вершинами $(-1, 0)$, $(0, -1)$, $(0, 1)$.
    *   Из равенства $|z + i| = |z - i|$ следует, что точка лежит на оси $Ox$ ($y = 0$).
    *   Из равенства $|z + 1| = |z - i|$: $\sqrt{(x+1)^2 + y^2} = \sqrt{x^2 + (y-1)^2}$. При $y=0$: $(x+1)^2 = x^2 + 1 \Rightarrow x^2 + 2x + 1 = x^2 + 1 \Rightarrow 2x = 0 \Rightarrow x = 0$.
*   **Что это:** Единственная точка $z = 0$ (начало координат).

### d) $0 < \arg \frac{i - z}{z + i} < \frac{\pi}{2}$
*   **Алгебра:** Преобразуем дробь: $\frac{i - z}{z + i} = \frac{-(z - i)}{z + i}$.
    *   Пусть $w = \frac{i - z}{z + i}$. Найдем его действительную и мнимую части.
    *   $w = \frac{-x + i(1-y)}{x + i(y+1)} = \frac{(-x + i(1-y))(x - i(y+1))}{x^2 + (y+1)^2} = \frac{(-x^2 + 1 - y^2) + i(2x)}{x^2 + (y+1)^2}$.
*   **Условие:** $0 < \arg w < \pi/2$ означает, что $\text{Re}(w) > 0$ и $\text{Im}(w) > 0$.
    *   $\text{Im}(w) > 0 \Rightarrow 2x > 0 \Rightarrow x > 0$ (правая полуплоскость).
    *   $\text{Re}(w) > 0 \Rightarrow 1 - x^2 - y^2 > 0 \Rightarrow x^2 + y^2 < 1$ (внутренность единичной окружности).
*   **Что это:** Пересечение правой полуплоскости и внутренности единичного круга.

### e) $-\frac{2\pi}{3} \leqslant \arg \frac{2 - iz}{z + 3i} \leqslant -\frac{5\pi}{6}$
*   **Алгебра:** Упростим выражение под аргументом:
    $\frac{2 - iz}{z + 3i} = \frac{-i(z + 2i)}{z + 3i} = -i \cdot \frac{z + 2i}{z + 3i}$.
    Тогда $\arg\left(-i \cdot \frac{z + 2i}{z + 3i}\right) = \arg(-i) + \arg(z + 2i) - \arg(z + 3i) = -\frac{\pi}{2} + \theta$, где $\theta = \arg(z + 2i) - \arg(z + 3i)$.
*   **Неравенство для $\theta$:**
    $-\frac{2\pi}{3} \leqslant -\frac{\pi}{2} + \theta \leqslant -\frac{5\pi}{6} \Rightarrow -\frac{\pi}{6} \leqslant \theta \leqslant -\frac{\pi}{3}$.
*   **Геометрический смысл:** $\theta$ — это угол, под которым виден отрезок между точками $-3i$ (точка $(0, -3)$) и $-2i$ (точка $(0, -2)$) из точки $z$.
    Угол отрицательный, значит точка $z$ находится справа от отрезка. Модуль угла меняется от $30^\circ$ до $60^\circ$.
*   **Что это:** Область, ограниченная двумя дугами окружностей, проходящих через точки $(0, -2)$ и $(0, -3)$. Это сегмент, расположенный справа от отрезка.
    *   Границы — дуги окружностей. Центры лежат на прямой $y = -2.5$.
    *   Радиусы: $R_1 = \frac{1}{2\sin(30^\circ)} = 1$ и $R_2 = \frac{1}{2\sin(60^\circ)} = \frac{1}{\sqrt{3}}$.

### f) $-\frac{2\pi}{3} \leqslant \arg \frac{3 - z}{z + i} \leqslant -\frac{\pi}{3}$
*   **Алгебра:** $\frac{3 - z}{z + i} = \frac{-(z - 3)}{z + i} = -1 \cdot \frac{z - 3}{z + i}$.
    $\arg = \arg(-1) + \arg(z - 3) - \arg(z + i) = \pi + \theta$, где $\theta = \arg(z - 3) - \arg(z + i)$.
*   **Неравенство для $\theta$:**
    $-\frac{2\pi}{3} \leqslant \pi + \theta \leqslant -\frac{\pi}{3} \Rightarrow -\frac{5\pi}{3} \leqslant \theta \leqslant -\frac{4\pi}{3}$.
    Прибавим $2\pi$: $\frac{\pi}{3} \leqslant \theta \leqslant \frac{2\pi}{3}$.
*   **Геометрический смысл:** $\theta$ — угол, под которым виден отрезок между точками $3$ (точка $(3, 0)$) и $-i$ (точка $(0, -1)$). Угол положительный, от $60^\circ$ до $120^\circ$.
*   **Что это:** Геометрическое место точек, из которых отрезок виден под углом от $60^\circ$ до $120^\circ$. Это "линза" (пересечение двух кругов), проходящая через точки $(3,0)$ и $(0,-1)$.
    *   Границы — две дуги окружностей, проходящих через эти точки.
    *   Радиус окружностей: $R = \frac{|AB|}{2\sin(60^\circ)} = \frac{\sqrt{10}}{\sqrt{3}}$.
